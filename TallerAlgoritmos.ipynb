1.Para determinar la orientacion de (A,B,C),Vamos a usar la formula del producto cruz,la cual equivale a la determinante de los dos vectores AB Y BC.
***USAMOS*** =$Det = (B_x - A_x) \cdot (C_y - B_y) - (B_y - A_y) \cdot (C_x - B_x)$

Decimos si $Det > 0$  : el giro es antihorario(**izquierda**)

Si , $Det < 0$:  el giro es horario(**Derecha**)

para $Det = 0$: Es colineal(siguen la misma linea)

Aplicando el procedimiento a los conjuntos de datos:Caso 1 (A: 1,2 | B: 5,3 | C: 3,7):$Det = (5 - 1) \cdot (7 - 3) - (3 - 2) \cdot (3 - 5)$$Det = (4 \cdot 4) - (1 \cdot -2) = 16 - (-2) = \mathbf{18}$Resultado: Giro antihorario.Caso 2 (A: 2,1 | B: 6,5 | C: 8,2):$Det = (6 - 2) \cdot (2 - 5) - (5 - 1) \cdot (8 - 6)$$Det = (4 \cdot -3) - (4 \cdot 2) = -12 - 8 = \mathbf{-20}$Resultado: Giro horario.Caso 3 (A: 1,1 | B: 3,3 | C: 6,6):$Det = (3 - 1) \cdot (6 - 3) - (3 - 1) \cdot (6 - 3)$$Det = (2 \cdot 3) - (2 \cdot 3) = 6 - 6 = \mathbf{0}$Resultado: Colineal.Caso 4 (A: 0,4 | B: 5,4 | C: 3,1):
$Det = (5 - 0) \cdot (1 - 4) - (4 - 4) \cdot (3 - 5)$
$Det = (5 \cdot -3) - (0 \cdot -2) = -15 - 0 = \mathbf{-15}$
Resultado: Giro horario.


```python
#1.3 implementacion en python

def calcular_orientacion(a, b, c):
    # a, b, c son tuplas o listas del tipo (x, y)
    det = (b[0] - a[0]) * (c[1] - b[1]) - (b[1] - a[1]) * (c[0] - b[0])

    if det > 0:
        return "Giro antihorario"
    elif det < 0:
        return "Giro horario"
    else:
        return "Colineal"
```

```python
def orientacion(p, q, r):
    det = (q[0] - p[0]) * (r[1] - q[1]) - (q[1] - p[1]) * (r[0] - q[0])
    if det > 0:
        return 1   # Antihorario
    elif det < 0:
        return -1  # Horario
    return 0       # Colineal

def en_segmento(p, q, r):
    # Verifica si el punto q está dentro del segmento pr
    return (min(p[0], r[0]) <= q[0] <= max(p[0], r[0]) and
            min(p[1], r[1]) <= q[1] <= max(p[1], r[1]))

def interseccion(a, b, c, d):
    o1 = orientacion(a, b, c)
    o2 = orientacion(a, b, d)
    o3 = orientacion(c, d, a)
    o4 = orientacion(c, d, b)

    # Caso general: cruce propio
    if o1 != o2 and o3 != o4:
        return True

    # Casos especiales: colineales con contacto o superposición
    if o1 == 0 and en_segmento(a, c, b):
        return True
    if o2 == 0 and en_segmento(a, d, b):
        return True
    if o3 == 0 and en_segmento(c, a, d):
        return True
    if o4 == 0 and en_segmento(c, b, d):
        return True

    return False
```

   4.4 Primera oreja mediante Ear ClippingUna primera oreja válida es el vértice $C(8,6)$, formando el triángulo $\triangle BCD$ (el vértice $E(0,6)$ con $\triangle DEA$ también es válido).¿Por qué $A$ y $B$ no son orejas válidas?    4.5 Condiciones antes de retirar un vértice (Ear Clipping)Antes de recortar un vértice $V_i$ (con vecinos $V_{i-1}$ y $V_{i+1}$), se deben verificar dos condiciones:Convexidad: El vértice $V_i$ debe ser convexo (su determinante de orientación debe ser positiva en sentido antihorario).Triángulo vacío: Ningún otro vértice del polígono (especialmente ningún vértice cóncavo) puede estar dentro ni sobre los bordes del triángulo $\triangle V_{i-1}V_iV_{i+1}$.

```python
def calcular_triangulacion(n):
    if n < 3:
        raise ValueError("Un polígono debe tener al menos 3 vértices.")

    triangulos = n - 2
    diagonales = n - 3
    return triangulos, diagonales

# Ejemplo de uso con el terreno de 5 vértices:
t, d = calcular_triangulacion(5)
print(f"Triángulos: {t}, Diagonales internas: {d}")
```
